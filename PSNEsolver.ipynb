```python
def solve_psne(p1_matrix, p2_matrix):
    n = len(p1_matrix)
    m = len(p1_matrix[0])
    p1_best_responses = set()
    for col in range(m):
        max_p1 = max(p1_matrix[row][col] for row in range(n))
        for row in range(n):
            if p1_matrix[row][col] == max_p1:
                p1_best_responses.add((row, col))
    p2_best_responses = set()
    for row in range(n):
        max_p2 = max(p2_matrix[row][col] for col in range(m))
        for col in range(m):
            if p2_matrix[row][col] == max_p2:
                p2_best_responses.add((row, col))
    return sorted(list(p1_best_responses.intersection(p2_best_responses)))
```
